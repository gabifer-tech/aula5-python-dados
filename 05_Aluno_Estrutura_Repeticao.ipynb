### Estruturas de Repetição em Python

As estruturas de repetição (ou loops) permitem executar um bloco de código várias vezes. Em Python, temos duas estruturas principais: **`for`** e **`while`**.

#### 1. O Loop `for`

O `for` é usado quando você quer iterar sobre uma sequência (como uma lista, tupla, dicionário, string) ou um intervalo numérico conhecido.

**Sintaxe:**
```python
for item in sequencia:
    # bloco de código a ser repetido
```

## Exemplo 1: Percorra um intervalo de números de 1 a 6 (usando range) e exiba apenas os números que são maiores que 3.

````python
for numero in range(1, 7):
    if numero > 3:
        print(f"O número {numero} é maior que 3.")
````

````python
# Exemplo 3: Iterando sobre uma lista de frutas
frutas = ['maçã', 'banana', 'uva']
for fruta in frutas:
    print(f"Eu gosto de {fruta}")

print("-" * 20)

# Exemplo 4: Usando a função range() para repetir 5 vezes (de 0 a 4)
for i in range(5):
    print(f"Contagem: {i}")
````

#### 2. O Loop `while`

O `while` executa um bloco de código enquanto uma determinada condição for verdadeira (`True`). É ideal quando você não sabe de antemão quantas vezes precisará repetir.

**Sintaxe:**
```python
while condicao:
    # bloco de código a ser repetido
```

````python
# Exemplo 1: Repetir até que o contador chegue a 5
contador = 0
while contador < 5:
    print(f"O contador atual é {contador}")
    contador += 1  # Incremento importante para evitar loop infinito
````

````python
# Exemplo 2: Crie um programa que peça para o usuário digitar uma palavra.
# O programa deve continuar pedindo novas palavras até que o usuário digite 'sair'.

palavra = ""

while palavra.lower() != "sair":
    palavra = input("Digite algo (ou 'sair' para encerrar): ")
    if palavra.lower() != "sair":
        print(f"Você digitou: {palavra}")

print("Programa encerrado!")
````

### Diferença entre `break` e `continue` no `while`

Tanto o `break` quanto o `continue` são instruções de controle de fluxo usadas para alterar o comportamento padrão de um loop (`while` ou `for`), mas eles funcionam de maneiras opostas:

---

#### **1. O comando `break`**
O `break` é usado para **interromper e sair imediatamente** do loop, independentemente de a condição do `while` ainda ser verdadeira ou não. Assim que o interpretador encontra o `break`, ele salta para a primeira linha de código fora do loop.

* **Exemplo de uso:** Interromper uma busca quando o elemento for encontrado.

```python
# Exemplo prático de 'break'
contador = 1
while contador <= 10:
    if contador == 5:
        print("Interrompendo o loop no 5 com o 'break'!")
        break  # Sai do loop imediatamente
    print(f"Contando: {contador}")
    contador += 1

print("Fora do loop!")
```
**Saída:**
```text
Contando: 1
Contando: 2
Contando: 3
Contando: 4
Interrompendo o loop no 5 com o 'break'!
Fora do loop!
```

---

#### **2. O comando `continue`**
O `continue` **não encerra** o loop. Em vez disso, ele **pula o restante do bloco de código** do ciclo atual e volta imediatamente para o teste da condição do `while` para iniciar a próxima repetição.

* **Atenção no `while`:** Ao usar `continue` em um loop `while`, certifique-se de atualizar a variável de controle (incrementar o contador) *antes* do `continue` para não criar um loop infinito.

```python
# Exemplo prático de 'continue' (pulando o número 3)
contador = 0
while contador < 5:
    contador += 1
    if contador == 3:
        print("Pulando o número 3 com o 'continue'...")
        continue  # Pula o print abaixo e volta para o teste do while
    print(f"Contando: {contador}")

print("Fora do loop!")
```
**Saída:**
```text
Contando: 1
Contando: 2
Pulando o número 3 com o 'continue'...
Contando: 4
Contando: 5
Fora do loop!
```

---

### **Resumo:**
* **`break`**: Para o loop por completo (encerra a repetição).
* **`continue`**: Pula apenas a iteração atual e vai para a próxima (continua repetindo).

````python
# ==========================================
# EXEMPLOS DE BREAK E CONTINUE NO LOOP FOR
# ==========================================

print("--- Exemplo 1: Usando 'break' no for ---")
# O loop vai de 1 a 5, mas será interrompido quando chegar no 4
for numero in range(1, 6):
    if numero == 4:
        print("  Encontramos o número 4! Parando o loop com 'break'.")
        break
    print(f"  Passando pelo número: {numero}")

print("\n" + "-"*40 + "\n")

print("--- Exemplo 2: Usando 'continue' no for ---")
# O loop vai de 1 a 5, mas vai pular o número 3 usando o 'continue'
for numero in range(1, 6):
    if numero == 3:
        print("  -> Pulando o número 3 com 'continue'...")
        continue
    print(f"  Passando pelo número: {numero}")
````
