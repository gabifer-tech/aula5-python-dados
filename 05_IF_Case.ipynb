### Condicionais `if` de Uma Linha (Operador Ternário e Inline)

Em Python, podemos simplificar estruturas de decisão simples para que ocupem apenas uma linha de código. Existem duas formas principais de fazer isso:

1. **Expressão Condicional (Operador Ternário):** Utilizado para atribuir valores ou retornar resultados com base em uma condição.
   - **Sintaxe:** `valor_se_verdadeiro if condicao else valor_se_falso`

2. **If Inline Simples:** Utilizado para executar uma instrução simples quando não há necessidade de um bloco `else`.
   - **Sintaxe:** `if condicao: instrucao`

```python
# ==========================================
# EXEMPLOS DE IF EM UMA LINHA
# ==========================================

# --- Exemplo 1: Operador Ternário (Atribuição de valor) ---
idade = 20
status = "Maior de idade" if idade >= 18 else "Menor de idade"
print(f"Exemplo 1: {status}")

# --- Exemplo 2: Operador Ternário direto no print ---
nota = 7.5
print(f"Exemplo 2: O aluno foi {'Aprovado' if nota >= 6 else 'Reprovado'}.")

# --- Exemplo 3: If Inline simples (sem else) ---
ativo = True
if ativo: print("Exemplo 3: O sistema está ativo!")
```

```python
# Exemplo de uso da estrutura match/case com guardas (if) em Python

# Vamos definir uma variável com a idade de uma pessoa
idade = 18

# Utilizamos o match para avaliar a variável idade
match idade:
    # Caso a idade seja menor que 0 (entrada inválida)
    case i if i < 0:
        print("Idade inválida!")

    # Caso a idade seja menor que 12
    case i if i < 12:
        print("Criança")

    # Caso a idade seja entre 12 e 17 anos
    case i if 12 <= i < 18:
        print("Adolescente")

    # Caso a idade seja maior ou igual a 18
    case i if i >= 18:
        print("Adulto")

    # Caso padrão (comportamento de fallback)
    case _:
        print("Idade desconhecida")
```
