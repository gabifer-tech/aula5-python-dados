# Exercícios de Python: Básico ao Avançado
Temas: Formatação, Estrutura Condicional, Match/Case e Repetição
### Restrições
- Não usar funções.
- Não usar try e except.

1. Leia o nome e a idade de uma pessoa e exiba:  
`Olá, João. Você tem 18 anos.`

2. Leia dois números inteiros e exiba:  
`A soma de X e Y é Z.`

3. Leia um número e informe se ele é positivo, negativo ou zero.

4. Leia 4 notas de um aluno.  
Se a média for maior ou igual a 7.5 → Aprovado  
Caso contrário → Reprovado

5. Leia um número inteiro e exiba sua tabuada de 1 a 10 usando for.

6. Leia um número inteiro.  
Informe se ele é:  
   - Par
   - Ou Ímpar

7. Leia três números e informe qual é o maior.

8. Leia um número de 1 a 7 e exiba o dia da semana correspondente utilizando match-case.  
Exemplo:  
1 → Domingo  
2 → Segunda-feira  
...

9. Leia uma senha.  
Enquanto a senha for diferente de "python123", solicite novamente.  
Ao acertar:  
Acesso permitido.  

10.  Leia um número inteiro positivo e calcule a soma de todos os números de 1 até ele.  
Exemplo:
     - 5
     - Resultado: 15  

11. Leia um número inteiro.  
Utilizando match-case, informe:  
    1 → Janeiro  
    2 → Fevereiro  
    ...  
    12 → Dezembro  
    Caso seja inválido:  
    Mês inválido.  

12. Leia um número inteiro.  
Conte quantos dígitos ele possui utilizando repetição.  
Exemplo:  
    - 12345  
    - Quantidade de dígitos: 5  

13. Leia um número inteiro positivo e informe se ele é primo.

14. Leia um número inteiro positivo e exiba todos os divisores dele.  
Exemplo:  
    - 12
    - 1 2 3 4 6 12

15. Leia dois números inteiros.  
Exiba o MDC (Máximo Divisor Comum) utilizando apenas estruturas de repetição e condicionais.
